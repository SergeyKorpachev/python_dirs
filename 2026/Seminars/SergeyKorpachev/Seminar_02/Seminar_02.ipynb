# Python, Семинар 2.

Поделаем некоторое количество задач по ООП.

## Простые и средние по сложности задачи.

**Задача 1: Класс Собака.**

Создай класс `Dog` с атрибутами `name` и `age`, и методом `bark()`, который печатает "Гав!". Создай объект и вызови метод.

**Задача 2: Класс Прямоугольник.**

Класс `Rectangle` с полями `width`, `height`. Методы `area()` и `perimeter()`. Добавь `__str__` для красивого вывода.

**Задача 3: Инкапсуляция и счётчик.**

Класс `Counter` с приватным полем `__value`. Методы `increment()`, `decrement()` и `get_value()`. Нельзя сделать значение отрицательным - бросай `ValueError`.

**Задача 4: Наследование - животные.**

Базовый класс `Animal` с методом `speak()`, возвращающим "...". Наследники `Cat` ("Мяу"), `Cow` ("Муу"). Функция `all_speak(animals)` вызывает `speak()` у каждого.

**Задача 5: Банковский счёт + история операций.**

Класс `Account` с балансом и списком операций. Методы `deposit(amount)`, `withdraw(amount)`. При снятии больше баланса - исключение `InsufficientFunds` (уже написано). Каждая операция пишется в историю. Реализуй `__str__` и `__eq__` (сравнение по номеру счёта).

# Python, Семинар 1

Поделаем некоторое количество задач от простых до более сложных.

## Простые задачи.

**Задача 1: Приветствие.**

Напишите программу, которая запрашивает имя пользователя и выводит приветствие.

**Задача 2: Сумма двух чисел.**

Пользователь вводит два числа, программа выводит их сумму.

**Задача 3: Четное или нечетное.**

Программа запрашивает число и определяет, четное оно или нет.

**Задача 4: Площадь прямоугольника.**

Пользователь вводит длину и ширину, программа вычисляет площадь.

**Задача 5: Сравнение двух чисел.**

Программа запрашивает два числа и выводит, какое больше (или они равны).

## Чуть сложнее задачи.

**Задача 6: Сумма чисел от 1 до N.**

Пользователь вводит N, программа считает сумму всех чисел от 1 до N.

**Задача 7: Таблица умножения.**

Программа выводит таблицу умножения для числа, введенного пользователем.

**Задача 8: Факториал.**

Пользователь вводит число N, программа вычисляет N! (факториал).

**Задача 9: Поиск максимума в списке.**

Пользователь вводит список чисел через пробел, программа находит максимум.

**Задача 10: Строка наоборот.**

Программа переворачивает введенную строку.

## Средний уровень задач.

**Задача 11: Простые числа.**

Программа выводит все простые числа до N, введенного пользователем.

**Задача 12: Количество слов в строке.**

Пользователь вводит строку, программа считает количество слов.

**Задача 13: Удаление дубликатов из списка.**

Программа принимает список и возвращает список с уникальными элементами.

**Задача 14: Палиндром.**

Программа проверяет, является ли введенная строка палиндромом.

**Задача 15: Частота символов.**

Программа подсчитывает, сколько раз каждый символ встречается в строке.

**Задача 16: Два списка в словарь.**

Программа принимает два списка (ключи и значения) и создает словарь.

**Задача 17: Слияние двух словарей.**

Программа объединяет два словаря (если ключи совпадают, значение берется из второго).

**Задача 18: Чтение файла и подсчет строк.**

Программа читает файл и выводит количество строк.

**Задача 19: Запись списка в файл.**

Программа записывает список строк в файл (каждая строка - новый элемент списка).

**Задача 20: Поиск подстроки в файле.**

Программа ищет все строки в файле, содержащие заданную подстроку.

```python
!ls output.txt && cat output.txt
```

## Продвинутые задачи.

**Задача 21: Генератор паролей.**

Программа генерирует случайный пароль заданной длины с использованием букв, цифр и символов.

**Задача 22: Транспонирование матрицы.**

Программа транспонирует матрицу (список списков).

**Задача 23: Сортировка словаря по значениям.**

Программа сортирует словарь по значениям (в порядке возрастания) и выводит результат.

**Задача 24: Класс для работы с дробями.**

Создайте класс Fraction для работы с обыкновенными дробями.

**Задача 25: Проверка на анаграммы.**

Две строки называются анаграммами, если они состоят из одних и тех же символов в разном порядке.

## Сложные задачи.

**Задача 26: Калькулятор с историей.**

Создайте класс Calculator, который хранит историю операций и поддерживает основные арифметические действия.

**Задача 27: Система управления библиотекой.**

Создайте классы Book и Library для управления книгами (добавление, удаление, поиск, выдача).

**Задача 28: Система учета студентов с оценками.**

Создайте программу для управления студентами и их оценками с использованием классов и файлового ввода-вывода.

## Не особо сложные задачи.

**Задача 29: Фильтрация вложенных структур.**

Напишите функцию, которая принимает вложенный список и условие-фильтр, и возвращает новый список с той же структурой, содержащий только элементы, удовлетворяющие условию (пустые вложенные списки удаляются).

**Задача 30: Группировка по ключам.**

Напишите функцию, которая принимает список элементов и функцию-ключ, и возвращает словарь, где ключи — это результаты применения функции к элементам, а значения — списки элементов, соответствующих этим ключам.

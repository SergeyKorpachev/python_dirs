**Задача 11: Простые числа.**

Программа выводит все простые числа до N, введенного пользователем.

```python
n = int(input("Введите N: "))  # 7

for num in range(2, n + 1):
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num, end=" ")
```

**Задача 15: Частота символов.**

```python
type({1, 2})
```

```python
type({1: 2})
```

```python
s = input("Введите строку: ")  # hello, world :)

freq = {}
for char in s:
    freq[char] = freq.get(char, 0) + 1
for char, count in freq.items():
    print(f"'{char}': {count}")
```

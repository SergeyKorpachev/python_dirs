```python
from typing import Union  #аннотирование типов
```

```python
class Complex:
    '''
    Комплексное число
    '''
    def __init__(self, re: Union[int, float] = 0, im: Union[int, float] = 0):
        self.re = re
        self.im = im

    def __str__(self) -> str:
        '''
        Метод
        '''
        if self.re == 0 and self.im == 0:
            return "0"
        elif self.im >= 0:
            return f"{self.re}+{self.im}i"
        else:
            return f"{self.re}-{abs(self.im)}i"

    __repr__ = __str__

    def __add__(self, other):
        '''
        Переопределяем сложение
        '''
        return self.__class__(self.re + other.re, self.im + other.im)

    def __sub__(self, other):
        return self.__class__(self.re - other.re, self.im - other.im)

    def __mul__(self, other):
        return self.__class__(
            self.re * other.re - self.im * other.im,
            self.re * other.im + self.im * other.re
        )

    def __eq__(self, other) -> bool:
        return self.re == other.re and self.im == other.im

    def __neg__(self):
        return self.__class__(-self.re, -self.im)

    # __radd__ = __add__
    # __rmul__ = __mul__
    # __rsub__ = __sub__
```

```python
class Number:
    def __init__(self, value):
        self.value = value

    # Handles: Number + other
    def __add__(self, other):
        if isinstance(other, Number):
            return Number(self.value + other.value)
        elif isinstance(other, (int, float)):
            return Number(self.value + other)
        return NotImplemented

    # Handles: other + Number (when 'other' doesn't know how to handle Number)
    def __radd__(self, other):
        # Addition is commutative (a + b == b + a),
        # so we can just reuse the __add__ logic!
        return self.__add__(other)

#####    __radd__ = __add__

    def __repr__(self):
        return f"Number({self.value})"


# --- Testing the code ---
n1 = Number(5)

# 1. Testing __add__ (Number + int)
print(n1 + 10)  # Output: Number(15)

# 2. Testing __radd__ (int + Number)
print(10 + n1)  # Output: Number(15)
```
